#Type checking


```python
l1 = [2,4,6,8,10]
var = pd.Series(l1)
print(var)
print(type(var))
```

#Inner

```python
Df1= pd.DataFrame( { 'A': [1,2,3,4], 'B': ['A','B','C','D']})
Df2= pd.DataFrame( { 'A': [1,2,3,4], 'C': ['a','b','c','d']})
pd.merge(Df1,Df2, on='A', how='inner', indicator=True)
```

#Outer


```python
Df1= pd.DataFrame( { 'A': [1,2,3,4], 'B': ['A','B','C','D']})
Df2= pd.DataFrame( { 'A': [1,2,3,4], 'C': ['a','b','c','d']})
pd.merge(Df1,Df2, on='A', how='outer',indicator=True)
```

#Left

```python
Df1= pd.DataFrame( { 'A': [1,2,3,4], 'B': ['A','B','C','D']})
Df2= pd.DataFrame( { 'A': [1,2,3,4], 'C': ['a','b','c','d']})
pd.merge(Df1,Df2, on='A', how='left',indicator=True)
```

#Righ

```python
Df1= pd.DataFrame( { 'A': [1,2,3,4], 'B': ['A','B','C','D']})
Df2= pd.DataFrame( { 'A': [1,2,3,4], 'C': ['a','b','c','d']})
pd.merge(Df1,Df2, on='A', how='right',indicator=True)
```

```python
Df1= pd.DataFrame( { 'A': [1,2,3,4], 'B': ['A','B','C','D']})
Df2= pd.DataFrame( { 'A': [1,5,7,3], 'C': ['a','b','c','d']})
pd.merge(Df1,Df2, on='A', how='right')
```

#SUFFIX

```python
Df1= pd.DataFrame( { 'A': [1,2,3,4,5], 'B': ['A','B','C','D','E']})
Df2= pd.DataFrame( { 'A': [1,2,3,4,5], 'B': ['a','b','c','d','e']})
pd.merge(Df1,Df2, left_index=True , right_index=True , suffixes=('_2019', '_2020'))
```

#INNER

```python

D1= pd.DataFrame( { 'A': [1,2,3,4,5], 'B': [10,20,30,40,50]}, index= ['a','b', 'c','d', 'e'])
D2= pd.DataFrame( { 'C': ['A','B'], 'D': ['E', 'F']}, index= ['c','d'])
D1.join(D2, how= 'inner')
```

##OUTER

```python

D1= pd.DataFrame( { 'A': [1,2,3,4,5], 'B': [10,20,30,40,50]}, index= ['a','b', 'c','d', 'e'])
D2= pd.DataFrame( { 'C': ['A','B'], 'D': ['E', 'F']}, index= ['c','d'])
D1.join(D2, how= 'outer')
```

##LEFT

```python

D1= pd.DataFrame( { 'A': [1,2,3,4,5], 'B': [10,20,30,40,50]}, index= ['a','b', 'c','d', 'e'])
D2= pd.DataFrame( { 'C': ['A','B'], 'D': ['E', 'F']}, index= ['c','d'])
D1.join(D2, how= 'left')
```

##RIGHT

```python

D1= pd.DataFrame( { 'A': [1,2,3,4,5], 'B': [10,20,30,40,50]}, index= ['a','b', 'c','d', 'e'])
D2= pd.DataFrame( { 'C': ['A','B'], 'D': ['E', 'F']}, index= ['c','d'])
D1.join(D2, how= 'right')
```

##JOINING SAME NAME TABLES

```python
D1= pd.DataFrame( { 'A': [1,2,3,4], 'B': ['A','B','C','D']})
D2= pd.DataFrame( { 'A': [1,2,3,4], 'B': ['a','b','c','d']})
D1.join(D2, lsuffix= '_left', rsuffix= '_right')
```
