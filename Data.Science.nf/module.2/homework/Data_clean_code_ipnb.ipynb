##Step 4: Identify missing values

```python
df.isnull().sum()
```

```python
df['Years of Experience'] = df['Years of Experience'].fillna(df['Years of Experience'].median())
df['Monthly Salary(INR)'] = df['Monthly Salary (INR)'].fillna(df['Monthly Salary (INR)'].median())
```

```python
Col=[
    'status','Age Group','Education','Indurstry','Location','AI Rist'
]
for col in df.columns:
    if df[col].dtype == "object":
        df[col] = df[col].fillna(df[col].mode()[0])
    else:
        df[col] = df[col].fillna(df[col].median())
```

```python
df.isnull().sum()
```

```python
df
```

##Step 5: Remove duplicate records

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated()
```

##Step 6: Correct inconsistent text formatting

```python
text_columns = df.select_dtypes(include="object").columns
for col in text_columns:
    df[col] = df[col].str.strip().str.title()
```

```python
df['Education'].value_counts()
```

```python
df['Education'] = df['Education'].replace({
    'Bachelor\'s': 'Bachelors',
    'Master': 'Masters',
    'Phd': 'PhD',
    'Ph.D': 'PhD'
})
```

```python
df["Age Group"] = df["Age Group"].astype(str).str.strip()

df["Age Group"] = df["Age Group"].replace({
    "18 to 24": "18-24",
    "25_34": "25-34",
    "35 - 44": "35-44"
})
```

```python
print(df["Age Group"].unique())
```

```python
df["Age Group"] = df["Age Group"].replace({
    "18 To 24": "18-24"
})
```

##Step 9: Rename and Standardize Column Names

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)
```
